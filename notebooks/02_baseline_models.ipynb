# 02. Baseline Models — TF-IDF & Logistic Regression
## AI Resume–Job Description Semantic Matching System

This notebook evaluates traditional lexical Information Retrieval (Cosine Similarity) and Classical Machine Learning (TF-IDF + Logistic Regression) baselines on candidate-job matching.


```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, str(Path.cwd().parent))
from src.config import PROCESSED_DATA_DIR, METRICS_DIR, SEED
from src.baseline import evaluate_tfidf_cosine_similarity, train_and_evaluate_logistic_regression
from src.dataset_builder import build_and_save_dataset

%matplotlib inline
```

### 1. Load Processed Dataset Splits


```python
train_path = PROCESSED_DATA_DIR / 'train.csv'
test_path = PROCESSED_DATA_DIR / 'test.csv'

if not train_path.exists() or not test_path.exists():
    train_df, val_df, test_df = build_and_save_dataset()
else:
    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)

print(f'Train Pairs: {len(train_df)} | Test Pairs: {len(test_df)}')
print('\nTrain Class Distribution:\n', train_df['label_name'].value_counts())
```

### 2. Baseline Model 1 — TF-IDF Cosine Similarity


```python
res_tfidf = evaluate_tfidf_cosine_similarity(train_df, test_df)
print('TF-IDF Cosine Similarity Results:')
for k, v in res_tfidf.items():
    if k != 'confusion_matrix':
        print(f'  {k}: {v}')
```

### 3. Baseline Model 2 — TF-IDF + Logistic Regression


```python
pipeline, res_logreg = train_and_evaluate_logistic_regression(train_df, test_df)
print('TF-IDF + Logistic Regression Results:')
for k, v in res_logreg.items():
    if k not in ['confusion_matrix', 'classification_report']:
        print(f'  {k}: {v}')
```
